```python
import os, re, json, csv
import pandas as pd
meta_path = GALAXY_INPUTS['meta'][0]['path']
meta = pd.read_csv(meta_path, sep='\t', index_col=0)
meta.index = [re.sub(r'[ \-]', '_', str(x)) for x in meta.index.astype(str)]
if not {'disease', 'cell_type'}.issubset(meta.columns):
    raise ValueError('Required metadata columns missing')
sheet = meta.reset_index().rename(columns={'index': 'sample'})
sheet = sheet[['sample', 'disease', 'cell_type']]
if sheet['sample'].duplicated().any():
    raise ValueError('Duplicate sample identifiers')
sheet.to_csv('outputs/sample_sheet.tsv', sep='\t', index=False, lineterminator='\n', quoting=csv.QUOTE_ALL)
print(json.dumps({'n_samples': int(sheet.shape[0]), 'columns': list(sheet.columns), 'samples': sheet['sample'].tolist()}))
```

```python
import os, h5py, numpy as np
src_path = GALAXY_INPUTS['source'][0]['path']
print('path', src_path)
with h5py.File(src_path, 'r') as f:
    print('root_keys', list(f.keys()))
    def show(name, obj):
        if name.count('/') <= 2:
            print('NODE', name, type(obj).__name__, dict(obj.attrs))
    f.visititems(show)
    for name in ['obs','var','layers','X']:
        if name in f:
            obj=f[name]
            print('TOP',name,type(obj).__name__, list(obj.keys())[:20] if hasattr(obj,'keys') else obj.shape)
```

```python
import os, json, numpy as np, pandas as pd
import anndata as ad
src_path = GALAXY_INPUTS['source'][0]['path']
adata = ad.read_h5ad(src_path)
print('shape', adata.shape)
print('obs_columns', list(adata.obs.columns))
print('var_columns', list(adata.var.columns))
print('var_name_head', [str(x) for x in adata.var_names[:10]])
for col in ['individual','disease','cell_type']:
    print(col, 'nunique', adata.obs[col].nunique(), 'values', [str(x) for x in adata.obs[col].astype(str).unique()[:30]])
print('layers', list(adata.layers.keys()))
print('counts_dtype', adata.layers['counts'].dtype, 'counts_type', type(adata.layers['counts']).__name__)
if 'gene_symbol' in adata.var.columns:
    print('gene_symbol_head', [str(x) for x in adata.var['gene_symbol'].iloc[:10]])
else:
    print('gene_symbol_column_absent')
print('individual_disease_consistency', adata.obs.groupby('individual', observed=True)['disease'].nunique().max())
```
